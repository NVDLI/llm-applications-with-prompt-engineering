## Setting Up the OpenAI Client

Paste the API key for your configured endpoint when the next cell prompts for it. The typed value is visible during entry but is not placed in the cell source or written to a file. After the setup cell runs, the key is available only to this notebook kernel. The following cell retrieves that in-memory value for the OpenAI client:

With a `base_url` and `api_key` we can now instantiate an OpenAI client. A bounded timeout and retry keep a delayed request from holding the notebook indefinitely.

```python
client = OpenAI(
    base_url=base_url,
    api_key=api_key,
    timeout=60.0,
    max_retries=1,
)
```

## Observing Available Models

Now that we've created an OpenAI client, we can use `client.models.list()` to inspect the available model catalog. We will keep the first view compact, then select the model used in this workshop.

```python
available_models = client.models.list()
```

```python
len(available_models.data)
```

The full response contains more model cards than we need to display. The following filter keeps the workshop models visible while leaving `available_models.data` available for further exploration:

```python
course_model = 'nvidia/nemotron-3-super-120b-a12b'
[model_card.id for model_card in available_models.data if model_card.id == course_model]
```

## Making a Simple Chat Completion Request

With the `client` instance now created, we can make a simple request to generate chat completions by using the `client.chat.completions.create` method which expects a `model` to use for the completion, as well as a list of `messages` to send to the model. We will be discussing the details of the `messages` list in more detail below, but for now we will pass in a simple single message containing a prompt from the user (you) asking for a fun fact about space. Next, inspect the returned completion and extract its answer.

```python
model = 'nvidia/nemotron-3-super-120b-a12b'
prompt = 'Tell me a fun fact about space.'
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[{'role': 'user', 'content': prompt}],
)
```

```python
print(response)
```

There's a fair amount of information provided in the API response, but the part we are most interested in is the response from the model.

Here we parse just the model's generated response out of the full API response.

```python
model_response = response.choices[0].message.content
```

```python
print(model_response)
```

## Understanding Completion and Chat Completion Endpoints

We have been working with the OpenAI-compatible `chat.completions` interface. The OpenAI client also defines a legacy `completions` interface, but it expects a plain-text prompt and a compatible completion model. The NVIDIA model used here is a chat model, so we use `chat.completions`.

The `chat.completions` interface accepts a list of role-labeled messages. The API does not retain a conversation for you; multi-turn context comes from sending the earlier messages again in that list.

The legacy `completions` interface accepts a plain prompt and predicts a continuation. It is not the interface exposed for the course model.

The main takeaway is that when working with OpenAI-compatible chat models such as `nvidia/nemotron-3-super-120b-a12b`, use `chat.completions` and not legacy `completions`.
