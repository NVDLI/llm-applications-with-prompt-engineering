**This section is the beginning of grand endeavor into programmatically working with large language models.** You'll start with the most basic of considerations like model types and interfaces to get comfortable. Gradually, you'll start to build towards more complex prompts, survey the vast surface area of helpful tooling, and get familiar with powerful LLM workflows with the help of LangChain. By the end of the course, you'll start to leverage powerful techniques for significant productivity leverage and will have good directions to follow as you continue to develop your skills in the space.

## Section Table of Contents

1. **NVIDIA NIM for Prompt Engineering:** In this notebook, we introduce NVIDIA NIMs and explore their capabilities throughout the course.
2. **Hello World with the OpenAI Library:** In this notebook, you will use the OpenAI client with the NVIDIA API Catalog to generate chat completions using Nemotron 3.5 Lightning.
3. **Hello World with LangChain:** In this notebook, we will learn how to interact with LangChain to generate chat completions using the same model.
4. **Streaming and Batching:** In this notebook you'll learn how to stream model responses and handle multiple chat completion requests in batches.
5. **Iterative Prompt Development:** In this notebook, you will learn the importance of iterating on prompts to achieve the desired responses from an LLM and explore how to write prompts that are specific.
6. **Prompt Templates:** In this notebook you'll learn how to capture reusable LLM functionality in prompt templates, and begin working with the powerful prompt template tools provided by LangChain.
